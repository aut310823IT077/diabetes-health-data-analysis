```python

import pandas as pd

df = pd.read_csv(
    "/content/diabetes_binary_health_indicators_BRFSS2015.csv"
)

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

display(df.head())

# Check the dataset size
print("Dataset shape:", df.shape)

# Display all column names
print("\nColumn names:")
print(df.columns.tolist())

# Check data types and non-null counts
print("\nDataset information:")
df.info()

# Get basic statistical summaries
print("\nStatistical summary:")
display(df.describe().T)

# Count missing values in each column
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nTotal missing values:", df.isnull().sum().sum())

# Count duplicate rows
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)

# Calculate duplicate percentage
duplicate_percentage = (duplicate_count / len(df)) * 100

print(f"Duplicate percentage: {duplicate_percentage:.2f}%")

# Count records in each diabetes group
print(df["Diabetes_binary"].value_counts().sort_index())

# Calculate percentages
print("\nPercentage distribution:")
print(
    (df["Diabetes_binary"].value_counts(normalize=True).sort_index() * 100)
    .round(2)
)

import matplotlib.pyplot as plt

diabetes_counts = df["Diabetes_binary"].value_counts().sort_index()

diabetes_counts.plot(kind="bar")

plt.title("Distribution of Diabetes Status")
plt.xlabel("Diabetes Status (0 = No, 1 = Yes)")
plt.ylabel("Number of Survey Responses")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Verify the current dataset
print("Current dataset shape:", df.shape)

print("\nDiabetes target counts:")
print(df["Diabetes_binary"].value_counts().sort_index())

print("\nDuplicate rows:", df.duplicated().sum())
print("Missing values:", df.isnull().sum().sum())
```

```python

# Recreate the cleaned dataset

df_original = df.copy()
df_clean = df_original.drop_duplicates().copy()

print("Original rows:", len(df_original))
print("Cleaned rows:", len(df_clean))
print("Duplicates removed:", len(df_original) - len(df_clean))
import sqlite3

# Create SQLite database
conn = sqlite3.connect("diabetes_health_analysis.db")

# Load cleaned data into SQL
df_clean.to_sql(
    "health_data",
    conn,
    if_exists="replace",
    index=False
)

print("Data successfully loaded into SQLite!")
query = """
SELECT *
FROM health_data
LIMIT 5;
"""

result = pd.read_sql_query(query, conn)

display(result)
query = """
SELECT COUNT(*) AS total_records
FROM health_data;
"""

result = pd.read_sql_query(query, conn)

display(result)
query = """
SELECT
    Diabetes_binary,
    COUNT(*) AS total_people
FROM health_data
GROUP BY Diabetes_binary
ORDER BY Diabetes_binary;
"""

result = pd.read_sql_query(query, conn)

display(result)
query = """
SELECT
    Diabetes_binary,
    COUNT(*) AS total_people,
    ROUND(
        COUNT(*) * 100.0 /
        (SELECT COUNT(*) FROM health_data),
        2
    ) AS percentage
FROM health_data
GROUP BY Diabetes_binary
ORDER BY Diabetes_binary;
"""

result = pd.read_sql_query(query, conn)

display(result)
```

```python
query = """
SELECT
    Diabetes_binary,
    HighBP,
    COUNT(*) AS total_people
FROM health_data
GROUP BY Diabetes_binary, HighBP
ORDER BY Diabetes_binary, HighBP;
"""

result = pd.read_sql_query(query, conn)

display(result)

query = """
SELECT
    Diabetes_binary,
    COUNT(*) AS total_people,
    SUM(HighBP) AS people_with_high_bp,
    ROUND(AVG(HighBP) * 100, 2) AS high_bp_percentage
FROM health_data
GROUP BY Diabetes_binary
ORDER BY Diabetes_binary;
"""

bp_result = pd.read_sql_query(query, conn)
display(bp_result)

query = """
SELECT
    Diabetes_binary,
    COUNT(*) AS total_people,
    ROUND(AVG(BMI), 2) AS average_bmi,
    MIN(BMI) AS minimum_bmi,
    MAX(BMI) AS maximum_bmi
FROM health_data
GROUP BY Diabetes_binary
ORDER BY Diabetes_binary;
"""

bmi_result = pd.read_sql_query(query, conn)
display(bmi_result)

query = """
SELECT
    Diabetes_binary,
    COUNT(*) AS total_people,
    ROUND(AVG(PhysActivity) * 100, 2) AS physically_active_percentage
FROM health_data
GROUP BY Diabetes_binary
ORDER BY Diabetes_binary;
"""

activity_result = pd.read_sql_query(query, conn)
display(activity_result)

query = """
SELECT
    Diabetes_binary,
    GenHlth,
    COUNT(*) AS total_people
FROM health_data
GROUP BY Diabetes_binary, GenHlth
ORDER BY Diabetes_binary, GenHlth;
"""

health_result = pd.read_sql_query(query, conn)
display(health_result)

query = """
SELECT
    Age,
    COUNT(*) AS total_people,
    ROUND(AVG(Diabetes_binary) * 100, 2) AS diabetes_percentage
FROM health_data
GROUP BY Age
ORDER BY Age;
"""

age_result = pd.read_sql_query(query, conn)
display(age_result)

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.bar(
    age_result["Age"].astype(str),
    age_result["diabetes_percentage"]
)

plt.title("Diabetes Percentage by Age Category")
plt.xlabel("Age Category Code")
plt.ylabel("Diabetes Percentage (%)")
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))

plt.bar(
    ["No diabetes", "Diabetes"],
    bp_result["high_bp_percentage"]
)

plt.title("High Blood Pressure by Diabetes Status")
plt.xlabel("Diabetes Status")
plt.ylabel("Respondents with High BP (%)")

for i, value in enumerate(bp_result["high_bp_percentage"]):
    plt.text(i, value + 1, f"{value:.1f}%", ha="center")

plt.tight_layout()
plt.savefig("high_bp_comparison.png", dpi=300, bbox_inches="tight")
plt.show()

plt.figure(figsize=(7, 5))

plt.bar(
    ["No diabetes", "Diabetes"],
    bmi_result["average_bmi"]
)

plt.title("Average BMI by Diabetes Status")
plt.xlabel("Diabetes Status")
plt.ylabel("Average BMI")

for i, value in enumerate(bmi_result["average_bmi"]):
    plt.text(i, value + 0.3, f"{value:.2f}", ha="center")

plt.tight_layout()
plt.savefig("bmi_comparison.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python

sql_script = """
-- Project: Diabetes Health Indicators Analysis
-- Database: SQLite
-- Purpose: Explore associations between health indicators
-- and diabetes status in the survey dataset.

-- Query 1: Total records
SELECT COUNT(*) AS total_records
FROM health_data;

-- Query 2: Diabetes group distribution
SELECT
    Diabetes_binary,
    COUNT(*) AS total_people,
    ROUND(AVG(Diabetes_binary) * 100, 2) AS percentage
FROM health_data
GROUP BY Diabetes_binary
ORDER BY Diabetes_binary;

-- Query 3: High blood pressure by diabetes status
SELECT
    Diabetes_binary,
    COUNT(*) AS total_people,
    SUM(HighBP) AS people_with_high_bp,
    ROUND(AVG(HighBP) * 100, 2) AS high_bp_percentage
FROM health_data
GROUP BY Diabetes_binary
ORDER BY Diabetes_binary;

-- Query 4: Average BMI by diabetes status
SELECT
    Diabetes_binary,
    COUNT(*) AS total_people,
    ROUND(AVG(BMI), 2) AS average_bmi
FROM health_data
GROUP BY Diabetes_binary
ORDER BY Diabetes_binary;

-- Query 5: Physical activity by diabetes status
SELECT
    Diabetes_binary,
    COUNT(*) AS total_people,
    ROUND(AVG(PhysActivity) * 100, 2)
        AS physically_active_percentage
FROM health_data
GROUP BY Diabetes_binary
ORDER BY Diabetes_binary;

-- Query 6: Diabetes percentage by age category
SELECT
    Age,
    COUNT(*) AS total_people,
    ROUND(AVG(Diabetes_binary) * 100, 2)
        AS diabetes_percentage
FROM health_data
GROUP BY Age
ORDER BY Age;

-- Query 7: General health by diabetes status
SELECT
    Diabetes_binary,
    GenHlth,
    COUNT(*) AS total_people
FROM health_data
GROUP BY Diabetes_binary, GenHlth
ORDER BY Diabetes_binary, GenHlth;
"""

with open("health_analysis.sql", "w", encoding="utf-8") as file:
    file.write(sql_script)

print("SQL script saved successfully!")
```
